```python
import pandas as pd
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/MaxWarriner/Data-Mining/refs/heads/main/imdb_top_1000.csv")
df = df.fillna(0)
df
```

```python
df = df[['Series_Title', 'IMDB_Rating', 'Meta_score', 'Gross', 'Certificate', 'Runtime']]
df['Runtime'] = df['Runtime'].str.replace('min', '')
df['Gross'] = df['Gross'].str.replace(',', '')
df['Gross'] = df['Gross'].astype(float)
df
```

```python
df['Runtime'] = df['Runtime'].astype(float)
```

```python
dummies = pd.get_dummies(df['Certificate'])
df = pd.concat([df, dummies], axis=1)
```

```python
df = df[['Series_Title', 'IMDB_Rating','Meta_score', 'Gross', 'Runtime', 'PG-13', 'R']]
```

```python
df
```

```python
df['PG-13'] = df['PG-13'].replace({True: 1, False: 0})
df
```

```python
df['R'] = df['R'].replace({True: 1, False: 0})
df
```

```python
from sklearn import preprocessing
import numpy as np
```

```python
scalingObj = preprocessing.MinMaxScaler()
new_df = scalingObj.fit_transform(df[['IMDB_Rating', 'Meta_score', 'Gross', 'Runtime', 'PG-13', 'R']])
new_df
```

```python
def min_finder(new_point, df):
  new_point_scaled = scalingObj.transform(new_point)
  df_diff = new_df - new_point_scaled
  df_diff = abs(df_diff)
  distances = df_diff.sum(axis=1).sort_values()
  print(distances.head(1).index)
```
