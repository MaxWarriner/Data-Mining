## Part 2 – Train a first model

Do the following:

1. Split the dataset into training and test sets using:
   - `test_size=0.3`
   - `random_state=42`
2. Train the following model:`KNeighborsRegressor`
3. Scale the features before training.
4. Fit the model on the training set.

**Short written answer:**

```python
# Write your code for Part 2 here
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsRegressor
from sklearn.preprocessing import MinMaxScaler

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

scaler = MinMaxScaler()
scaler.fit(X_train)

X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)

neigh = KNeighborsRegressor(n_neighbors=2)
neigh.fit(X_train, y_train)
```

**Short answer for Part 2:**

We used the K-nearest-neighbors regression model. This model is computationally efficient when a new data point is added because no re-training of a larger model is required. Feature scaling is needed because otherwise features with larger scales would dominate the distance calculations.

```python
# Write your code for Part 3 here
preds = neigh.predict(X_test)
MAE = abs(preds - y_test).sum() / len(y_test)
print(MAE)
baseline_MAE = abs(y_train.mean() - y_test).sum() / len(y_test)
print(baseline_MAE)
```
